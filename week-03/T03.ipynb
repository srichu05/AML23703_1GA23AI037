```python
import sys
!{sys.executable} -m pip install qiskit

from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

qc = QuantumCircuit(1)

gate_sequence = [
    ("H", qc.h),
    ("S", qc.s),
    ("Y", qc.y),
    ("T", qc.t),
    ("Z", qc.z)
]

print("Applied Gates:")

for name, gate in gate_sequence:
    print(name, end=" ")
    gate(0)

state = Statevector.from_instruction(qc)

print("\n\nCircuit")
print(qc.draw())

print("\nStatevector")
print(state)
```
