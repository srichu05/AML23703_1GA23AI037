```python
!pip install qiskit

from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

def display_gate(gate):
    qc = QuantumCircuit(1)

    if gate == "X":
        qc.x(0)
    elif gate == "Y":
        qc.y(0)
    elif gate == "Z":
        qc.z(0)

    print(f"\n----- {gate} Gate -----")
    print(qc.draw())
    print(Statevector.from_instruction(qc))

for gate in ["X", "Y", "Z"]:
    display_gate(gate)
```
