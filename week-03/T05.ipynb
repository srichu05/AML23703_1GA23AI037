```python
import sys
!{sys.executable} -m pip install qiskit
from qiskit import QuantumCircuit
from qiskit.quantum_info import Operator
import numpy as np

c1 = QuantumCircuit(1)
c1.h(0)
c1.z(0)
c1.h(0)

c2 = QuantumCircuit(1)
c2.x(0)

matrix1 = Operator(c1).data
matrix2 = Operator(c2).data

print("HZH Matrix")
print(matrix1)

print("\nX Matrix")
print(matrix2)

print("\nEquivalent :", np.allclose(matrix1, matrix2))
```
