```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
import numpy as np

# Create circuit
qc = QuantumCircuit(2)

# Apply Hadamard to both qubits
qc.h(0)
qc.h(1)

# Get statevector
state = Statevector.from_instruction(qc)

print("Quantum Circuit:")
print(qc)

print("\nStatevector:")
print(state)

# Display amplitudes
print("\nAmplitudes:")
for i, amp in enumerate(state.data):
    print(f"|{i:02b}> : {amp}")

# Verify normalization
probability = np.sum(np.abs(state.data)**2)

print("\nSum of probabilities =", probability)
```
